## Ejercicio 5
*(Exercise 5)*
Realiza un conversor de grados Celsius a Fahrenheit. El conversor tiene que recibir el input del usuario y devolver la conversión.

*(Create a converter from degrees Celsius to Fahrenheit. The converter must receive input from the user and return the conversion.)*

La conversión de grados sigue la siguiente fórmula: F = (C × 9 / 5) + 32
*(The degree conversion follows the formula: F = (C × 9 / 5) + 32)*

```python
# Ask the user to enter a temperature in Celsius
celsius = float(input("Enter temperature in Celsius: "))
# Apply the conversion formula F = (C * 9/5) + 32
fahrenheit = (celsius * 9 / 5) + 32
print("El equivalente en grados Fahrenheit es", fahrenheit)
```
