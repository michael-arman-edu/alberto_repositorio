#### Ejercicio 3

Crea una **función que reciba por parámetro** una lista de strings. Los strings pueden valer "rojo", "verde", "amarillo".

La función debe devolver otra lista de strings, donde si el string original era "rojo" la salida debe ser "stop", si era "verde" la salida debe ser "avance" y si era "amarillo" debe ser "precaución".

---
*(Exercise 3)*

*Create a **function that receives a list of strings** as a parameter. The strings can be "rojo", "verde", or "amarillo".*

*The function must return a new list where "rojo" → "stop", "verde" → "avance", "amarillo" → "precaución"*

```python
def semaforo(colores):
    resultado = []
    for color in colores:
        if color == "rojo":
            resultado.append("stop")
        elif color == "verde":
            resultado.append("avance")
        elif color == "amarillo":
            resultado.append("precaución")
    return resultado

print(semaforo(["rojo", "verde", "amarillo", "rojo"]))
```

Crea un programa que simule un carrusel infantil.

El carrusel infantil tiene 3 posibles cochecitos: barco - bomberos - avioneta (una variable global de tipo lista, fuera de tu función).

Debes escribir una **FUNCIÓN** llamada carrusel que tenga dos parámetros de entrada: el cochecito que eliges (puede valer el string barco, bomberos o avioneta) y un segundo parámetro llamado vueltas que es un entero.

Tu función carrusel NO debe retornar nada, pero debe llamar internamente a coloca(cochecito), ya te lo doy programado. Si sientas al niño en "avioneta" y pones 3 vueltas, la función debería imprimir:

"vuelta 1: avioneta - barco - bomberos"

"vuelta 3: avioneta - barco - bomberos"

Se valorará positivamente que haya **UN BUCLE WHILE**.
Si no te acaba de funcionar llamando a coloca(cochecito) tras un tiempo prudencial, escribe tu código sin usarlo.

---
*(Exercise 4 — Children's Carousel)*

*Simulate a children's carousel with 3 rides: barco, bomberos, avioneta (a global list).*

*Write a **FUNCTION** called carrusel with two parameters: the chosen ride (string) and vueltas (int, number of rounds).*

*The function returns nothing but calls coloca(cochecito) internally (already provided). For "avioneta" with 3 rounds it should print:*

*"vuelta 1: avioneta - barco - bomberos"* *(repeated for each round)*

*A **WHILE LOOP** is encouraged.*

```python
lista_carrusel = ["barco", "bomberos", "avioneta"]

def coloca(cochecito):
    if cochecito == "barco":
        return lista_carrusel
    elif cochecito == "bomberos" or cochecito == "avioneta":
        elemento = lista_carrusel.pop(0)
        lista_carrusel.append(elemento)
    if cochecito == "avioneta":
        elemento = lista_carrusel.pop(0)
        lista_carrusel.append(elemento)
    return lista_carrusel

def carrusel(cochecito, vueltas):
    vuelta = 1
    while vuelta <= vueltas:
        orden = coloca(cochecito)
        print(f"vuelta {vuelta}: {' - '.join(orden)}")
        vuelta += 1

carrusel("avioneta", 3)
```

Crea un programa que pida una contraseña por teclado e indique si es correcta
o incorrecta con un print(). La contraseña correcta es "python". Si la contraseña era incorrecta la pide de nuevo hasta 3 veces como máximo. Si al tercer intento no se acierta, debe ponerse a None y poner en un print() que la cuenta se ha bloqueado. Si se consigue entrar, se muestra que es correcta en un print(). Debe haber como mínimo un **BUCLE WHILE**

---
*(Exercise 5 — Password checker)*

*Ask the user for a password and say if it is correct or incorrect. The correct password is "python".*

*If wrong, ask again up to 3 times maximum.*

*If the third attempt fails, set the password variable to None and print that the account is blocked.*

*If correct, print that it is correct. Must include at least one **WHILE LOOP**.*

```python
contrasena_correcta = "python"
intentos = 0
contrasena = None

while intentos < 3:
    contrasena = input("Introduce la contraseña: ")
    if contrasena == contrasena_correcta:
        print("Contraseña correcta")
        break
    else:
        intentos += 1
        print("Contraseña incorrecta")

if intentos == 3:
    contrasena = None
    print("La cuenta se ha bloqueado")
```

*(Exercise 6 — Flower class)*

*Create a class Flor (Flower) where all attributes are set via the constructor:*

*nombre → string (flower name)*

*venenosa → boolean (is it poisonous?)*

*lista_viveros → list of strings (nurseries where it can be bought)*

*Methods:*

*cantidad_de_viveros → no parameters, returns the number of nurseries (int)*

*imprime_info(nombre) → no return value, prints: "Hola nombre, la flor X es/no es venenosa y se puede comprar en N viveros"*

*The constructor **MUST** have default values. Create a default flower and print its info.*

```python
class Flor:
    def __init__(self, nombre="rosa", venenosa=False, lista_viveros=["vivero1"]):
        self.nombre = nombre
        self.venenosa = venenosa
        self.lista_viveros = lista_viveros

    def cantidad_de_viveros(self):
        return len(self.lista_viveros)

    def imprime_info(self, nombre):
        es_o_no = "es" if self.venenosa else "no es"
        print(f"Hola {nombre}, la flor {self.nombre} {es_o_no} venenosa y se puede comprar en {self.cantidad_de_viveros()} viveros")

flor_defecto = Flor()
flor_defecto.imprime_info("Alberto")
```

*Using map and the built-in reversed (at minimum):*

*Write a program that receives a list of strings and returns a tuple with each string reversed. **NO LOOPS ALLOWED**.*

*Example: ["hola", "adiós"] → ("aloh", "sóida")*

*Hint: use a string method that joins characters, and you may need more than one map!*
