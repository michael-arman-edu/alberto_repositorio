#### Ejercicio 2. 

Escribe un programa que te diga si una letra es una consonante o una vocal. Crea tu propia **FUNCIÓN**

---
*(Exercise 2 — Vowel or consonant)*

*Write a program that tells whether a letter is a vowel or a consonant. Create your own **FUNCTION**.*

````python
def vocal_o_consonante(letra):
    vocales = "aeiouaeiou"
    if letra.lower() in vocales:
        print(f"{letra} es una vocal")
    else:
        print(f"{letra} es una consonante")

vocal_o_consonante("a")
vocal_o_consonante("b")
vocal_o_consonante("e")
vocal_o_consonante("z")
````

Escribe un programa que calcule la edad humana de un perro.
Tienes que pedirle al usuario la edad del perro en años.
Si el usuario introduce un número negativo, imprime que la cifra facilitada no tiene sentido. Si el perro tiene un año, son 14 años humanos. Si tiene 2 años, son 22 años humanos.
A partir de los 2 años, el cálculo es el siguiente: a la edad del perro se le restan 2 años, el resultado se multiplica por 5 y, finalmente, se le suma 22.
Usa una función sin argumentos de entrada ni de salida (muestra toda la información con print())

---
*(Exercise 3 — Dog age in human years)*

*Ask the user for the dog's age. If negative, print that it makes no sense.*

*1 dog year = 14 human years. 2 dog years = 22 human years.*

*For ages > 2: human_age = (dog_age - 2) * 5 + 22*

*Use a function with no input parameters and no return value (use print() for output).*

````python
def edad_perro():
    edad = int(input("Introduce la edad del perro en años: "))
    if edad < 0:
        print("La cifra facilitada no tiene sentido")
    elif edad == 1:
        print("El perro tiene 14 años humanos")
    elif edad == 2:
        print("El perro tiene 22 años humanos")
    else:
        humana = (edad - 2) * 5 + 22
        print(f"El perro tiene {humana} años humanos")

edad_perro()
````

#### Ejercicio 5. 

Escribe un programa que te diga si una frase es un pangrama. Un pangrama es una frase que contiene todas las letras del alfabeto.
Puedes tener espacios en blanco, pero NO tendrás signos de puntuación, a excepción de las tildes.

Ejemplo de pangrama:

Extraño pan de col y kiwi se quemó bajo fugaz vaho

---
*(Exercise 5 — Pangram checker)*

*Write a program that tells whether a sentence is a pangram (contains every letter of the alphabet).*

*Spaces are allowed but there will be no punctuation except accents (tildes).*

*Example pangram: Extraño pan de col y kiwi se quemó bajo fugaz vaho*

````python
def es_pangrama(frase):
    alfabeto = "abcdefghijklmnopqrstuvwxyzñ"
    frase_lower = frase.lower()
    return all(letra in frase_lower for letra in alfabeto)

frase1 = "Extraño pan de col y kiwi se quemo bajo fugaz vaho"
frase2 = "Hola mundo"
print(es_pangrama(frase1))
print(es_pangrama(frase2))
````

#### Ejercicio 7.

Usando objetos modela un theremin (es un instrumento musical que se toca modificando el campo electromagnético, por si tenías curiosidad).
Tendrá de atributos la marca, el modelo, el peso, la potencia y la batería.

Todos los instrumentos se crean con un peso predeterminado de 500, una potencia de 50 y una batería de 200.

Cuando se recarga la batería vale 200.

Cuando se toca, se muestra por pantalla que está sonando y la batería disminuye 10 (no siendo nunca negativo).
Si no hay batería, no puede sonar. Crea un instrumento, tócalo mientras quede batería, cárgalo y no lo vuelvas a tocar.

---
*(Exercise 7 — Theremin class)*

*Model a theremin using OOP. Attributes: marca, modelo, peso (default 500), potencia (default 50), bateria (default 200).*

*Recharging sets bateria back to 200.*

*Playing prints that it is sounding and reduces bateria by 10 (never below 0). If no battery, it cannot play.*

*Create an instrument, play it until the battery runs out, recharge it, then don't play it again.*

````python
class Theremin:
    def __init__(self, marca, modelo, peso=500, potencia=50, bateria=200):
        self.marca = marca
        self.modelo = modelo
        self.peso = peso
        self.potencia = potencia
        self.bateria = bateria

    def recargar(self):
        self.bateria = 200
        print("Batería recargada")

    def tocar(self):
        if self.bateria > 0:
            print(f"El theremin {self.modelo} está sonando. Batería: {self.bateria}")
            self.bateria -= 10
        else:
            print("Sin batería, no puede sonar")

t = Theremin("Moog", "Etherwave")

while t.bateria > 0:
    t.tocar()

t.recargar()
````

#### Ejercicio 8
Una empresa usa dos listas para guardar la información de sus empleados. Una lista guarda el nombre del empleado y la otra guarda su salario. Escribe un programa que cree esas dos listas originalmente vacías y que a través de un menú pueda hacer lo siguiente:
    
Inserta 1 para añadir un nuevo empleado y su salario

Inserta 2 para imprimir los nombres y salarios de todos los empleados

Inserta 3 para mostrar el número de empleados

Inserta 4 para imprimir los nombres de los empleados con sueldos superiores a 400000

Inserta 5 para subir un 5% los sueldos por debajo de 10000

Inserta 6 para mostrar el total de todos los salarios

Inserta 7 para salir del programa

---
*(Exercise 8 — Employee menu)*

*A company uses two lists: one for employee names, one for salaries (both start empty).*

*Create a menu-driven program:*

*1 → add a new employee and salary*

*2 → print all names and salaries*

*3 → show the number of employees*

*4 → print names of employees earning more than 400,000*

*5 → raise by 5% all salaries below 10,000*

*6 → show total salary sum*

*7 → exit the program*

````python
nombres = []
salarios = []

print("Bienvenido al gestor de empleados")

while True:
    print("
1. Añadir empleado")
    print("2. Ver todos los empleados")
    print("3. Número de empleados")
    print("4. Empleados con sueldo > 400000")
    print("5. Subir 5% sueldos < 10000")
    print("6. Total de salarios")
    print("7. Salir")

    opcion = input("Elige una opción: ")

    if opcion == "1":
        nombre = input("Nombre del empleado: ")
        salario = float(input("Salario: "))
        nombres.append(nombre)
        salarios.append(salario)
        print(f"Empleado {nombre} añadido")
    elif opcion == "2":
        for nombre, salario in zip(nombres, salarios):
            print(f"{nombre}: {salario}")
    elif opcion == "3":
        print(f"Número de empleados: {len(nombres)}")
    elif opcion == "4":
        for nombre, salario in zip(nombres, salarios):
            if salario > 400000:
                print(nombre)
    elif opcion == "5":
        for i in range(len(salarios)):
            if salarios[i] < 10000:
                salarios[i] *= 1.05
        print("Salarios actualizados")
    elif opcion == "6":
        print(f"Total de salarios: {sum(salarios)}")
    elif opcion == "7":
        print("Saliendo...")
        break
    else:
        print("Opción no válida")
````

````python
lista_pelis_original = ["Toy Story", "Soul", "Up", "Coco", "Luca", "Monsters", "Inside Out", "WallE", "Buscando a Nemo", "Los increíbles"]          
````

````python
lista_pelis_original = ["Toy Story", "Soul", "Up", "Coco", "Luca", "Monsters", "Inside Out", "WallE", "Buscando a Nemo", "Los increibles"]

def favorita(lista, ganador):
    if not lista:
        print(f"
Prefieres finalmente {ganador}")
        return
    rival = lista[0]
    resto = lista[1:]
    eleccion = input(f"Si prefieres {ganador} marca 1, si prefieres {rival} marca 2: ")
    if eleccion == "1":
        print(f"Prefieres {ganador}")
        favorita(resto, ganador)
    else:
        print(f"Prefieres {rival}")
        favorita(resto, rival)

pelis = lista_pelis_original.copy()
primera = pelis[0]
resto = pelis[1:]
favorita(resto, primera)
````
