## Ejercicio 1
*(Exercise 1)*
Modelizar una tienda de electrodomésticos como clase con atributos, métodos y proyección de ventas.
*(Model an appliance store as a class with attributes, methods and sales projection.)*

```python
class Tienda:
    tipo = "Electrodomésticos"
    abierta = True

    def __init__(self, nombre, direccion, num_empleados, ventas_3_meses):
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_3_meses = ventas_3_meses

    def total_ventas(self):
        return sum(self.ventas_3_meses)

    def media_ventas_empleado(self):
        return self.total_ventas() / self.num_empleados

    def info(self):
        return f"{self.nombre} - {self.direccion}"

    def ventas_ultimo_mes(self):
        return self.ventas_3_meses[-1]

    def proyeccion_marketing(self, x):
        factor = 1.2 if x < 1000 else 1.5
        return [v * factor for v in self.ventas_3_meses]
```

```python
tienda1 = Tienda("ElectroMax", "Calle Mayor 10", 15, [20000, 25000, 30000])
tienda2 = Tienda("TechStore", "Avenida Central 5", 10, [15000, 18000, 22000])
tienda3 = Tienda("CasaHogar", "Avenida del Parque 20", 8, [12000, 14000, 16000])

print("Atributos de clase:")
print("tipo:", Tienda.tipo)
print("abierta:", Tienda.abierta)

print("\nAtributos de instancia de tienda1:")
print("nombre:", tienda1.nombre)
print("direccion:", tienda1.direccion)
print("num_empleados:", tienda1.num_empleados)
print("ventas_3_meses:", tienda1.ventas_3_meses)

print("\nMétodos de tienda1:")
print("total_ventas:", tienda1.total_ventas())
print("media_ventas_empleado:", tienda1.media_ventas_empleado())
print("info:", tienda1.info())
print("ventas_ultimo_mes:", tienda1.ventas_ultimo_mes())
print("proyeccion_marketing(500):", tienda1.proyeccion_marketing(500))
print("proyeccion_marketing(2000):", tienda1.proyeccion_marketing(2000))
```

```python
tiendas = [tienda1, tienda2, tienda3]

total_ultimo_mes = sum(t.ventas_ultimo_mes() for t in tiendas)
print("Total ventas último mes:", total_ultimo_mes)

print("\nTiendas en Avenida:")
for t in tiendas:
    if "Avenida" in t.direccion:
        print(t.nombre)
```

## Ejercicio 2
*(Exercise 2)*
Implementar una clase Perro con atributos fijos, métodos de movimiento y ladrido.
*(Implement a Dog class with fixed attributes, movement and bark methods.)*

```python
class Perro:
    patas = 4
    orejas = 2
    ojos = 2

    def __init__(self, raza, pelo="Marrón", dueño=None):
        self.raza = raza
        self.pelo = pelo
        self.dueño = dueño
        self.velocidad = 0

    def andar(self, aumento_velocidad):
        self.velocidad += aumento_velocidad

    def parar(self):
        self.velocidad = 0

    def ladrar(self, mensaje):
        return "GUAU!, " + mensaje
```

```python
perro = Perro("Labrador")

print("Atributos de instancia:")
print("raza:", perro.raza)
print("pelo:", perro.pelo)
print("dueño:", perro.dueño)
print("velocidad:", perro.velocidad)

print("\nAtributos de clase:")
print("patas:", Perro.patas)
print("orejas:", Perro.orejas)
print("ojos:", Perro.ojos)

perro.andar(10)
print("\nVelocidad tras andar(10):", perro.velocidad)

perro.parar()
print("Velocidad tras parar():", perro.velocidad)

print("\n" + perro.ladrar("¿Quién eres?"))
```
