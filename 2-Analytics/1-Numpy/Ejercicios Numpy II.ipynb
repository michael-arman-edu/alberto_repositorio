## Ejercicio 9
*(Exercise 9)*

Crea una matriz aleatoria de 5x5 y estandariza sus valores. Para normalizar tienes que restar sus valores por la media(`np.mean`) y dividir el resultado por la desviación estándar(`np.std`).
*(Create a random 5x5 matrix and standardise its values. To normalise you have to subtract the mean (`np.mean`) and divide the result by the standard deviation (`np.std`).)*

````python
# create a random 5*5 matrix and standardize its values
# to normalize you have to subtrat the mean(np.mean) &
# divide by the standard deviation (np.std)

matrix = np.random.rand(5, 5)
matrix_standardized = (matrix - np.mean(matrix)) / np.std(matrix)
print(matrix_standardized)
````

## Ejercicio 10
*(Exercise 10)*

1. Crea una secuencia de números que vayan del 0 al 10.
   *(Create a sequence of numbers from 0 to 10.)*
2. Crea un nuevo array, que parta del anterior, pero sustituya todos los elementos entre el 3 y el 8, por el valor -1
   *(Create a new array, starting from the previous one, but replacing all elements between 3 and 8 with the value -1)*
3. Crea un nuevo array que niegue todos los elementos entre el 3 y el 8
   *(Create a new array that negates all elements between 3 and 8)*

````python
# create a sequence of numbers from 0 to 10
array_1 = np.arange(0, 11)
print(array_1)

# create a new array, starting from pervious one but replacing all elements
# between 3 and 8 with -1

array_2 = np.where((array_1 >= 3) & (array_1 <= 8), -1, array_1)
print(array_2)

````
